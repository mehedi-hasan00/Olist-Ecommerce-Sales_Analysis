```python
import pandas as pd
import numpy as np
```

```python
data_path = 'olist_dataset/'
orders = pd.read_csv(data_path+'olist_orders_dataset.csv')
order_items = pd.read_csv(data_path+'olist_order_items_dataset.csv')
customers = pd.read_csv(data_path+'olist_customers_dataset.csv')
products = pd.read_csv(data_path+'olist_products_dataset.csv')
payments = pd.read_csv(data_path+'olist_order_payments_dataset.csv')
reviews = pd.read_csv(data_path+'olist_order_reviews_dataset.csv')
category_translation = pd.read_csv(data_path+'product_category_name_translation.csv')
```

```python
for name, df_ in [('orders',orders),('order_items',order_items),('customers',customers),('products',products),('payments',payments),('reviews',reviews)]:
    print(f"{name}:{df_.shape}")
```

#### All are in tring datatype we have to change the data type string to datetime

```python
date_cols = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]
for col in date_cols:
    orders[col] = pd.to_datetime(orders[col])
```

### Product dataset

```python
products.head()
```

#### Product category name are in Portuguese language, we have to translate it into English Language

```python
products_clean = products.merge(
    category_translation,
    on = 'product_category_name',
    how= 'left'
)
```

```python
products_clean.head()
```

```python
products_clean.drop(columns = ['product_category_name'], inplace = True)
products_clean.rename(columns={'product_category_name_english':'product_category_name'}, inplace = True)
```

```python
products_clean.head()
```

```python
products_clean.isnull().sum()
```

```python
products_clean['product_category_name'] = products_clean['product_category_name'].fillna('unknown')
```

```python
products_clean.isnull().sum()
```

```python
products_clean = products_clean.dropna()
```

```python
products_clean.isnull().sum()
```
